```python
import numpy as np
from scipy.stats import norm

def power(n, d, alpha=0.05):
    z = norm.ppf(1 - alpha / 2)
    return norm.cdf(d * np.sqrt(n / 2) - z)
```

```python
ns = np.arange(0, 2001, 20)
fig, ax = plt.subplots()
for d in (0.1, 0.2, 0.3):
    ax.plot(ns, power(ns, d), label=f"d = {d}")
ax.axhline(0.8, color="#A8A194", linestyle="--", linewidth=1)
ax.set_xlabel("users per group"); ax.set_ylabel("power"); ax.legend(frameon=False)
plt.show()
```
